```python

from pathlib import Path
import pandas as pd

print("HYDROGEL MACHINE LEARNING RESEARCH")
print("=" * 50)

# Find experimental Excel files
files = list(Path.cwd().rglob("*.xlsx"))

for file in files:
    print(f"\nDataset: {file.name}")

    workbook = pd.ExcelFile(file)

    for sheet in workbook.sheet_names:
        df = pd.read_excel(file, sheet_name=sheet)

        print(f"\nSheet: {sheet}")
        print(f"Shape: {df.shape}")
        print("Columns:", df.columns.tolist())

        print("\nFirst 5 rows:")
        print(df.head().to_string(index=False))
```

```python

from pathlib import Path
import pandas as pd

# Locate the PEGDA workbook
files = list(Path.cwd().rglob(
    "PEGDA_Characterization_Data.xlsx"
))

if not files:
    raise FileNotFoundError(
        "PEGDA workbook not found."
    )

file = files[0]

# Read without assuming a header row
df_raw = pd.read_excel(
    file,
    sheet_name="PEGDA_3",
    header=None
)

print("Dataset shape:", df_raw.shape)

# Show the first 25 rows and 15 columns
display(df_raw.iloc[:25, :15])

```
