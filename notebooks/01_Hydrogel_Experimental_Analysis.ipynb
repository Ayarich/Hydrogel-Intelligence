```python

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Select one compression workbook
folder = Path.cwd() / "PEGDA_CompressionFiles"
filepath = folder / "10pPEGDA_60s_Compression.xlsx"

workbook = pd.ExcelFile(filepath)

print("Available sheets:", workbook.sheet_names)

# Start with the first worksheet
raw = pd.read_excel(
    filepath,
    sheet_name=0,
    header=None
)

# Measurements begin at Excel row 8
data = raw.iloc[7:, 0:6].copy()

data.columns = [
    "load_N",
    "time_s",
    "extension_mm",
    "corrected_extension_mm",
    "strain",
    "stress_Pa"
]

# Convert measurements to numerical values
for column in data.columns:
    data[column] = pd.to_numeric(
        data[column],
        errors="coerce"
    )

# Remove rows without stress or strain
data = data.dropna(subset=["strain", "stress_Pa"])

# Select the 1–5% strain region
linear_region = data[
    (data["strain"] >= 0.01) &
    (data["strain"] <= 0.05)
].copy()

print("Total measurements:", len(data))
print("Measurements in 1–5% region:", len(linear_region))

if len(linear_region) < 3:
    raise ValueError(
        "Insufficient measurements in the selected strain range."
    )

# Train a linear regression model
X = linear_region[["strain"]]
y = linear_region["stress_Pa"]

model = LinearRegression()
model.fit(X, y)

# Slope of stress versus strain is apparent modulus
modulus_Pa = model.coef_[0]
modulus_kPa = modulus_Pa / 1000

print(f"Apparent Young's modulus: {modulus_kPa:.3f} kPa")
print(f"Fit R²: {model.score(X, y):.4f}")

# Plot experimental measurements
plt.figure(figsize=(9, 5))

plt.scatter(
    data["strain"] * 100,
    data["stress_Pa"] / 1000,
    s=12,
    alpha=0.5,
    label="Experimental data"
)

# Plot fitted line only over the selected region
x_fit = np.linspace(0.01, 0.05, 100)
y_fit = model.predict(
    pd.DataFrame({"strain": x_fit})
)

plt.plot(
    x_fit * 100,
    y_fit / 1000,
    linewidth=2,
    label=f"Linear fit: E = {modulus_kPa:.2f} kPa"
)

plt.xlabel("Compressive strain (%)")
plt.ylabel("Compressive stress (kPa)")
plt.title("Hydrogel Compression: Stress–Strain Analysis")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.linear_model import LinearRegression

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

workbook = pd.ExcelFile(filepath)

results = []

for sheet in workbook.sheet_names:

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    # Select experimental measurements
    data = raw.iloc[7:, 0:6].copy()

    data.columns = [
        "load_N",
        "time_s",
        "extension_mm",
        "corrected_extension_mm",
        "strain",
        "stress_Pa"
    ]

    for column in data.columns:
        data[column] = pd.to_numeric(
            data[column],
            errors="coerce"
        )

    data = data.dropna(
        subset=["strain", "stress_Pa"]
    )

    # Select the reported fitting interval
    region = data[
        (data["strain"] >= 0.01)
        & (data["strain"] <= 0.05)
    ].copy()

    if len(region) < 3:
        print(f"Insufficient data: {sheet}")
        continue

    model = LinearRegression()

    model.fit(
        region[["strain"]],
        region["stress_Pa"]
    )

    modulus_kPa = model.coef_[0] / 1000

    results.append({
        "Specimen": sheet,
        "Measurements": len(data),
        "Fit points": len(region),
        "Modulus (kPa)": modulus_kPa,
        "R2": model.score(
            region[["strain"]],
            region["stress_Pa"]
        )
    })

results_df = pd.DataFrame(results)

display(results_df)

print("\nMean modulus (kPa):",
      results_df["Modulus (kPa)"].mean())

print("Standard deviation (kPa):",
      results_df["Modulus (kPa)"].std(ddof=1))
```

```python

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LinearRegression

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

reported = {
    "PEGDA_1": 21.409,
    "PEGDA_2": 27.954,
    "PEGDA_3": 28.051
}

comparison_results = []

for sheet in pd.ExcelFile(filepath).sheet_names:

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    data = raw.iloc[7:, 0:6].copy()
    data.columns = [
        "load_N",
        "time_s",
        "extension_mm",
        "corrected_extension_mm",
        "strain",
        "stress_Pa"
    ]

    for col in data.columns:
        data[col] = pd.to_numeric(
            data[col],
            errors="coerce"
        )

    data = data.dropna(
        subset=["strain", "stress_Pa"]
    )

    # Restrict to 1–5% strain
    region = data[
        data["strain"].between(0.01, 0.05)
    ].copy()

    if len(region) < 3:
        continue

    X = region[["strain"]]
    y = region["stress_Pa"]

    # Method 1: Ordinary linear regression
    model = LinearRegression()
    model.fit(X, y)

    ordinary = model.coef_[0] / 1000

    # Method 2: Regression through the origin
    origin_model = LinearRegression(
        fit_intercept=False
    )
    origin_model.fit(X, y)

    through_origin = origin_model.coef_[0] / 1000

    # Method 3: Secant between points
    # closest to 1% and 5% strain
    low_idx = (region["strain"] - 0.01).abs().idxmin()
    high_idx = (region["strain"] - 0.05).abs().idxmin()

    low = region.loc[low_idx]
    high = region.loc[high_idx]

    secant = (
        (high["stress_Pa"] - low["stress_Pa"])
        /
        (high["strain"] - low["strain"])
    ) / 1000

    comparison_results.append({
        "Specimen": sheet,
        "Reported": reported.get(sheet),
        "Linear regression": ordinary,
        "Through origin": through_origin,
        "Secant modulus": secant
    })

method_df = pd.DataFrame(comparison_results)

display(method_df.round(3))
```

```python

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

results = []

for sheet in pd.ExcelFile(filepath).sheet_names:

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    df = raw.iloc[7:, 0:6].copy()

    df.columns = [
        "load_N", "time_s",
        "extension_mm", "corrected_extension_mm",
        "strain", "stress_Pa"
    ]

    df["strain"] = pd.to_numeric(
        df["strain"], errors="coerce"
    )

    df["stress_Pa"] = pd.to_numeric(
        df["stress_Pa"], errors="coerce"
    )

    df = df.dropna(subset=["strain", "stress_Pa"])

    region = df[df["strain"].between(0.01, 0.05)]

    X = region[["strain"]]
    y = region["stress_Pa"]

    ordinary = LinearRegression()
    ordinary.fit(X, y)

    origin = LinearRegression(fit_intercept=False)
    origin.fit(X, y)

    rmse_ordinary = np.sqrt(
        mean_squared_error(y, ordinary.predict(X))
    )

    rmse_origin = np.sqrt(
        mean_squared_error(y, origin.predict(X))
    )

    results.append({
        "Specimen": sheet,
        "Intercept (Pa)": ordinary.intercept_,
        "Ordinary slope (kPa)": ordinary.coef_[0] / 1000,
        "Origin slope (kPa)": origin.coef_[0] / 1000,
        "Ordinary RMSE (Pa)": rmse_ordinary,
        "Origin RMSE (Pa)": rmse_origin
    })

diagnostics = pd.DataFrame(results)

display(diagnostics.round(3))
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.linear_model import LinearRegression

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

intervals = [
    (0.01, 0.03),
    (0.01, 0.05),
    (0.02, 0.05),
    (0.03, 0.07)
]

results = []

workbook = pd.ExcelFile(filepath)

for sheet in workbook.sheet_names:

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    data = raw.iloc[7:, 0:6].copy()

    data.columns = [
        "load_N",
        "time_s",
        "extension_mm",
        "corrected_extension_mm",
        "strain",
        "stress_Pa"
    ]

    for col in ["strain", "stress_Pa"]:
        data[col] = pd.to_numeric(
            data[col],
            errors="coerce"
        )

    data = data.dropna(
        subset=["strain", "stress_Pa"]
    )

    for low, high in intervals:

        region = data[
            data["strain"].between(low, high)
        ]

        if len(region) < 3:
            continue

        model = LinearRegression()

        model.fit(
            region[["strain"]],
            region["stress_Pa"]
        )

        results.append({
            "Specimen": sheet,
            "Strain interval": f"{low*100:.0f}-{high*100:.0f}%",
            "Modulus (kPa)": model.coef_[0] / 1000,
            "R²": model.score(
                region[["strain"]],
                region["stress_Pa"]
            ),
            "Points": len(region)
        })

sensitivity_df = pd.DataFrame(results)

display(sensitivity_df.round(3))

# Compare results visually
pivot = sensitivity_df.pivot(
    index="Strain interval",
    columns="Specimen",
    values="Modulus (kPa)"
)

pivot.plot(marker="o", figsize=(9, 5))

plt.ylabel("Apparent compressive modulus (kPa)")
plt.xlabel("Strain fitting interval")
plt.title("Sensitivity of PEGDA Modulus to Fitting Interval")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

fig, ax = plt.subplots(figsize=(9, 6))
summary = []

for sheet in pd.ExcelFile(filepath).sheet_names:

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    data = raw.iloc[7:, 0:6].copy()
    data.columns = [
        "load_N", "time_s", "extension_mm",
        "corrected_extension_mm", "strain", "stress_Pa"
    ]

    for col in ["strain", "stress_Pa"]:
        data[col] = pd.to_numeric(
            data[col],
            errors="coerce"
        )

    data = data.dropna(subset=["strain", "stress_Pa"])

    # Restrict modeling to initial compression
    region = data[data["strain"].between(0.01, 0.10)].copy()

    X = region[["strain"]]
    y = region["stress_Pa"] / 1000  # kPa

    poly = PolynomialFeatures(degree=2, include_bias=False)
    X_poly = poly.fit_transform(X)

    model = LinearRegression()
    model.fit(X_poly, y)

    # Stress = a*strain² + b*strain + c
    b, a = model.coef_
    c = model.intercept_

    strain_grid = np.linspace(0.01, 0.10, 100)
    grid_df = pd.DataFrame({"strain": strain_grid})
    y_pred = model.predict(poly.transform(grid_df))

    tangent_2 = 2 * a * 0.02 + b
    tangent_5 = 2 * a * 0.05 + b
    tangent_8 = 2 * a * 0.08 + b

    summary.append({
        "Specimen": sheet,
        "Fit R²": model.score(X_poly, y),
        "Tangent E at 2% (kPa)": tangent_2,
        "Tangent E at 5% (kPa)": tangent_5,
        "Tangent E at 8% (kPa)": tangent_8
    })

    ax.scatter(
        region["strain"] * 100,
        y,
        s=8,
        alpha=0.3
    )

    ax.plot(
        strain_grid * 100,
        y_pred,
        linewidth=2,
        label=f"{sheet} quadratic fit"
    )

ax.set_xlabel("Compressive strain (%)")
ax.set_ylabel("Compressive stress (kPa)")
ax.set_title("Nonlinear PEGDA Compression Modeling")
ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()

display(pd.DataFrame(summary).round(3))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

filepath = (
    Path.cwd()
    / "PEGDA_CompressionFiles"
    / "10pPEGDA_60s_Compression.xlsx"
)

evaluation_results = []

fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

for ax, sheet in zip(
    axes,
    pd.ExcelFile(filepath).sheet_names
):

    raw = pd.read_excel(
        filepath,
        sheet_name=sheet,
        header=None
    )

    data = raw.iloc[7:, 0:6].copy()

    data.columns = [
        "load_N",
        "time_s",
        "extension_mm",
        "corrected_extension_mm",
        "strain",
        "stress_Pa"
    ]

    for column in ["strain", "stress_Pa"]:
        data[column] = pd.to_numeric(
            data[column],
            errors="coerce"
        )

    data = data.dropna(
        subset=["strain", "stress_Pa"]
    )

    data = data[
        data["strain"].between(0.01, 0.10)
    ].sort_values("strain")

    X = data[["strain"]].to_numpy()
    y = data["stress_Pa"].to_numpy() / 1000

    split = int(len(X) * 0.70)

    X_train = X[:split]
    X_test = X[split:]

    y_train = y[:split]
    y_test = y[split:]

    # Linear model
    linear = LinearRegression()
    linear.fit(X_train, y_train)

    linear_pred = linear.predict(X_test)

    # Quadratic model
    poly = PolynomialFeatures(
        degree=2,
        include_bias=False
    )

    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)

    quadratic = LinearRegression()
    quadratic.fit(X_train_poly, y_train)

    quadratic_pred = quadratic.predict(X_test_poly)

    for name, predictions in [
        ("Linear", linear_pred),
        ("Quadratic", quadratic_pred)
    ]:

        evaluation_results.append({
            "Specimen": sheet,
            "Model": name,
            "MAE (kPa)": mean_absolute_error(
                y_test,
                predictions
            ),
            "RMSE (kPa)": np.sqrt(
                mean_squared_error(
                    y_test,
                    predictions
                )
            ),
            "R2": r2_score(
                y_test,
                predictions
            )
        })

    ax.scatter(
        X_train[:, 0] * 100,
        y_train,
        s=12,
        label="Training data"
    )

    ax.scatter(
        X_test[:, 0] * 100,
        y_test,
        s=12,
        label="Test data"
    )

    ax.plot(
        X_test[:, 0] * 100,
        linear_pred,
        label="Linear prediction"
    )

    ax.plot(
        X_test[:, 0] * 100,
        quadratic_pred,
        label="Quadratic prediction"
    )

    ax.set_title(sheet)
    ax.set_xlabel("Strain (%)")
    ax.set_ylabel("Stress (kPa)")
    ax.grid(alpha=0.25)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()

evaluation_df = pd.DataFrame(evaluation_results)

display(evaluation_df.round(4))
```

```python

from pathlib import Path
import re
import pandas as pd
from sklearn.linear_model import LinearRegression

folder = Path.cwd() / "GelMA_CompressionFiles"

records = []

for filepath in sorted(folder.glob("*.xlsx")):

    filename = filepath.name

    # Extract concentration and curing time
    match = re.search(
        r"(\d+)p(\d+)s",
        filename,
        flags=re.IGNORECASE
    )

    if not match:
        print("Skipping filename:", filename)
        continue

    concentration = float(match.group(1))
    curing_time = int(match.group(2))

    workbook = pd.ExcelFile(filepath)

    for sheet in workbook.sheet_names:

        raw = pd.read_excel(
            filepath,
            sheet_name=sheet,
            header=None
        )

        # Find the actual measurement header
        header_rows = raw.index[
            raw.iloc[:, 0].astype(str)
               .str.strip()
               .eq("Load (N)")
        ]

        if len(header_rows) == 0:
            print("No measurement header:", filename, sheet)
            continue

        start = header_rows[0] + 2

        data = raw.iloc[start:, 0:6].copy()

        data.columns = [
            "load_N",
            "time_s",
            "extension_mm",
            "corrected_extension_mm",
            "strain",
            "stress_Pa"
        ]

        for col in ["strain", "stress_Pa"]:
            data[col] = pd.to_numeric(
                data[col],
                errors="coerce"
            )

        data = data.dropna(
            subset=["strain", "stress_Pa"]
        )

        # Fit the predefined 1–5% strain interval
        region = data[
            data["strain"].between(0.01, 0.05)
        ].copy()

        if len(region) < 3:
            print("Insufficient points:", filename, sheet)
            continue

        model = LinearRegression()
        model.fit(
            region[["strain"]],
            region["stress_Pa"]
        )

        records.append({
            "filename": filename,
            "specimen": sheet,
            "GelMA_concentration_pct": concentration,
            "curing_time_s": curing_time,
            "modulus_kPa": model.coef_[0] / 1000,
            "fit_R2": model.score(
                region[["strain"]],
                region["stress_Pa"]
            ),
            "fit_points": len(region)
        })

dataset = pd.DataFrame(records)

if dataset.empty:
    raise ValueError(
        "No valid compression measurements were extracted."
    )

dataset = dataset.sort_values([
    "GelMA_concentration_pct",
    "curing_time_s",
    "specimen"
]).reset_index(drop=True)

display(dataset)

print("\nTotal specimen records:", len(dataset))
print("Unique formulations:",
      dataset[[
          "GelMA_concentration_pct",
          "curing_time_s"
      ]].drop_duplicates().shape[0])

# Save processed data separately from original files
output_dir = Path.cwd() / "data" / "processed"
output_dir.mkdir(parents=True, exist_ok=True)

dataset.to_csv(
    output_dir / "gelma_modulus_dataset.csv",
    index=False
)

print("\nSaved processed dataset successfully.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the formulation-level summary
df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
)

features = [
    "GelMA_concentration_pct",
    "curing_time_s"
]

target = "mean_modulus_kPa"

X = df[features]
y = df[target]

# Three candidate models
models = {
    "Mean baseline": DummyRegressor(strategy="mean"),

    "Linear regression": make_pipeline(
        StandardScaler(),
        LinearRegression()
    ),

    "Quadratic Ridge": make_pipeline(
        StandardScaler(),
        PolynomialFeatures(
            degree=2,
            include_bias=False
        ),
        Ridge(alpha=1.0)
    )
}

cv = LeaveOneOut()
results = []
predictions = pd.DataFrame({
    "concentration": X["GelMA_concentration_pct"],
    "curing_time": X["curing_time_s"],
    "actual_kPa": y
})

for name, model in models.items():

    # Each prediction is made by a model that
    # did not see that formulation during training.
    y_pred = cross_val_predict(
        model, X, y, cv=cv
    )

    mae = mean_absolute_error(y, y_pred)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    r2 = r2_score(y, y_pred)

    results.append({
        "Model": name,
        "MAE (kPa)": mae,
        "RMSE (kPa)": rmse,
        "LOOCV R2": r2
    })

    predictions[name] = y_pred

results_df = pd.DataFrame(results)
results_df = results_df.sort_values("RMSE (kPa)")

print("MODEL COMPARISON")
display(results_df.round(3))

print("\nHELD-OUT FORMULATION PREDICTIONS")
display(predictions.round(3))

# Actual vs predicted plot
fig, axes = plt.subplots(
    1, 3, figsize=(15, 4.5)
)

for ax, (name, model) in zip(axes, models.items()):
    ax.scatter(
        y,
        predictions[name],
        s=60
    )

    limits = [
        min(y.min(), predictions[name].min()),
        max(y.max(), predictions[name].max())
    ]

    ax.plot(limits, limits, "--")
    ax.set_xlabel("Actual modulus (kPa)")
    ax.set_ylabel("Predicted modulus (kPa)")
    ax.set_title(name)
    ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.compose import TransformedTargetRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
)

X = df[[
    "GelMA_concentration_pct",
    "curing_time_s"
]]

y = df["mean_modulus_kPa"]

if (y <= 0).any():
    raise ValueError("Log modeling requires positive targets.")

def quadratic_model():
    return make_pipeline(
        StandardScaler(),
        PolynomialFeatures(
            degree=2,
            include_bias=False
        ),
        Ridge(alpha=1.0)
    )

models = {
    "Ordinary Quadratic Ridge": quadratic_model(),

    "Log-Target Quadratic Ridge":
        TransformedTargetRegressor(
            regressor=quadratic_model(),
            func=np.log,
            inverse_func=np.exp,
            check_inverse=True
        )
}

cv = LeaveOneOut()

results = []
prediction_table = df[[
    "GelMA_concentration_pct",
    "curing_time_s",
    "mean_modulus_kPa"
]].copy()

for name, model in models.items():

    pred = cross_val_predict(
        model, X, y, cv=cv
    )

    results.append({
        "Model": name,
        "MAE (kPa)": mean_absolute_error(y, pred),
        "RMSE (kPa)": np.sqrt(
            mean_squared_error(y, pred)
        ),
        "LOOCV R2": r2_score(y, pred),
        "Minimum prediction (kPa)": pred.min(),
        "Negative predictions": int((pred < 0).sum())
    })

    prediction_table[name] = pred

print("MODEL EVALUATION")
display(pd.DataFrame(results).round(3))

print("\nFORMULATION PREDICTIONS")
display(prediction_table.round(3))

# Compare held-out predictions
plt.figure(figsize=(8, 6))

for name in models:
    plt.scatter(
        y,
        prediction_table[name],
        label=name,
        s=55
    )

limits = [
    min(0, prediction_table[list(models)].min().min()),
    max(y.max(), prediction_table[list(models)].max().max())
]

plt.plot(limits, limits, "k--", label="Perfect prediction")

plt.xlabel("Measured mean modulus (kPa)")
plt.ylabel("LOOCV predicted modulus (kPa)")
plt.title("GelMA: Ordinary vs Positive-Constrained Regression")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    ConstantKernel, Matern, WhiteKernel
)
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
)

X = df[
    ["GelMA_concentration_pct", "curing_time_s"]
].to_numpy()

y = df["mean_modulus_kPa"].to_numpy()

assert np.all(y > 0)

loo = LeaveOneOut()

predictions = np.zeros(len(y))
lower_95 = np.zeros(len(y))
upper_95 = np.zeros(len(y))

for train_idx, test_idx in loo.split(X):

    X_train = X[train_idx]
    X_test = X[test_idx]

    # Standardize using training data only
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Model the logarithm of modulus
    y_train_log = np.log(y[train_idx])

    # Fixed kernel for our introductory experiment
    kernel = (
        ConstantKernel(1.0)
        * Matern(
            length_scale=[1.0, 1.0],
            nu=2.5
        )
        + WhiteKernel(noise_level=0.05)
    )

    model = GaussianProcessRegressor(
        kernel=kernel,
        optimizer=None,
        normalize_y=True,
        alpha=1e-8
    )

    model.fit(X_train_scaled, y_train_log)

    mean_log, std_log = model.predict(
        X_test_scaled,
        return_std=True
    )

    idx = test_idx[0]

    # Back-transform:
    # exp(mean_log) is the median on the original scale
    predictions[idx] = np.exp(mean_log[0])

    lower_95[idx] = np.exp(
        mean_log[0] - 1.96 * std_log[0]
    )

    upper_95[idx] = np.exp(
        mean_log[0] + 1.96 * std_log[0]
    )

# Evaluate out-of-fold predictions
mae = mean_absolute_error(y, predictions)
rmse = np.sqrt(mean_squared_error(y, predictions))
r2 = r2_score(y, predictions)

coverage = np.mean(
    (y >= lower_95) & (y <= upper_95)
)

print("GAUSSIAN PROCESS RESULTS")
print(f"MAE: {mae:.3f} kPa")
print(f"RMSE: {rmse:.3f} kPa")
print(f"LOOCV R²: {r2:.3f}")
print(f"Nominal 95% interval coverage: {coverage:.1%}")

results = pd.DataFrame({
    "GelMA_pct": X[:, 0],
    "curing_time_s": X[:, 1],
    "actual_kPa": y,
    "predicted_kPa": predictions,
    "lower_95_kPa": lower_95,
    "upper_95_kPa": upper_95
})

display(results.round(3))

# Visualize actual versus predicted modulus
plt.figure(figsize=(8, 6))

plt.errorbar(
    y,
    predictions,
    yerr=[
        predictions - lower_95,
        upper_95 - predictions
    ],
    fmt="o",
    capsize=3,
    label="GPR predictions"
)

max_value = max(y.max(), upper_95.max())

plt.plot(
    [0, max_value],
    [0, max_value],
    "--",
    label="Perfect prediction"
)

plt.xlabel("Measured modulus (kPa)")
plt.ylabel("Predicted modulus (kPa)")
plt.title("Gaussian Process Regression: GelMA Hydrogels")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
)

X = df[
    ["GelMA_concentration_pct", "curing_time_s"]
].to_numpy()

y = df["mean_modulus_kPa"].to_numpy()

std = df["std_modulus_kPa"].to_numpy()
n = df["specimen_count"].to_numpy()

if (
    np.any(y <= 0)
    or np.any(n < 2)
    or np.any(~np.isfinite(std))
):
    raise ValueError("Check target and replicate statistics.")

# Estimated standard error of each formulation mean
sem = std / np.sqrt(n)

# Delta-method approximation to variance of log(mean)
noise_variance_log = (sem / y) ** 2

# Small numerical floor, not a measured noise estimate
noise_variance_log = np.maximum(
    noise_variance_log, 1e-6
)

loo = LeaveOneOut()

pred = np.zeros(len(y))
lower = np.zeros(len(y))
upper = np.zeros(len(y))

for train_idx, test_idx in loo.split(X):

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X[train_idx])
    X_test = scaler.transform(X[test_idx])

    log_y_train = np.log(y[train_idx])

    kernel = (
        ConstantKernel(1.0)
        * Matern(
            length_scale=[1.0, 1.0],
            nu=2.5
        )
    )

    model = GaussianProcessRegressor(
        kernel=kernel,
        alpha=noise_variance_log[train_idx],
        optimizer=None,
        normalize_y=False
    )

    model.fit(X_train, log_y_train)

    mu, sigma = model.predict(
        X_test,
        return_std=True
    )

    idx = test_idx[0]

    pred[idx] = np.exp(mu[0])

    lower[idx] = np.exp(
        mu[0] - 1.96 * sigma[0]
    )

    upper[idx] = np.exp(
        mu[0] + 1.96 * sigma[0]
    )

metrics = {
    "MAE_kPa": mean_absolute_error(y, pred),
    "RMSE_kPa": np.sqrt(
        mean_squared_error(y, pred)
    ),
    "LOOCV_R2": r2_score(y, pred)
}

print("NOISE-AWARE GPR")
print(metrics)

comparison = df[[
    "GelMA_concentration_pct",
    "curing_time_s",
    "mean_modulus_kPa"
]].copy()

comparison["predicted_kPa"] = pred
comparison["lower_95_kPa"] = lower
comparison["upper_95_kPa"] = upper
comparison["SEM_kPa"] = sem
comparison["log_noise_variance"] = noise_variance_log

display(comparison.round(3))
```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import LeaveOneOut
from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv("data/processed/gelma_formulation_summary.csv")

X = df[["GelMA_concentration_pct", "curing_time_s"]].to_numpy()
y = df["mean_modulus_kPa"].to_numpy()
sd = df["std_modulus_kPa"].to_numpy()
n = df["specimen_count"].to_numpy()

assert np.all(y > 0)
assert np.all(np.isfinite(sd))
assert np.all(n >= 2)

# Provisional independent-replicate assumption
sem = sd / np.sqrt(n)
alpha_variable = np.maximum((sem / y) ** 2, 1e-6)

# Constant noise at the median estimated variance
alpha_constant = np.full(
    len(y), np.median(alpha_variable)
)

experiments = {
    "A: Zero mean + constant noise": (False, False),
    "B: Zero mean + variable noise": (False, True),
    "C: Centered + constant noise": (True, False),
    "D: Centered + variable noise": (True, True)
}

scores = []
prediction_table = df[[
    "GelMA_concentration_pct",
    "curing_time_s",
    "mean_modulus_kPa"
]].copy()

for name, (center_target, variable_noise) in experiments.items():

    predictions = np.zeros(len(y))

    for train_idx, test_idx in LeaveOneOut().split(X):

        scaler = StandardScaler()
        X_train = scaler.fit_transform(X[train_idx])
        X_test = scaler.transform(X[test_idx])

        log_y = np.log(y[train_idx])

        # Estimate prior mean from training data only
        offset = log_y.mean() if center_target else 0.0
        centered_y = log_y - offset

        alpha = (
            alpha_variable[train_idx]
            if variable_noise
            else alpha_constant[train_idx]
        )

        kernel = (
            ConstantKernel(1.0, constant_value_bounds="fixed")
            * Matern(
                length_scale=[1.0, 1.0],
                length_scale_bounds="fixed",
                nu=2.5
            )
        )

        gp = GaussianProcessRegressor(
            kernel=kernel,
            alpha=alpha,
            optimizer=None,
            normalize_y=False
        )

        gp.fit(X_train, centered_y)

        predicted_log = gp.predict(X_test)[0] + offset
        predictions[test_idx[0]] = np.exp(predicted_log)

    scores.append({
        "Experiment": name,
        "MAE_kPa": mean_absolute_error(y, predictions),
        "RMSE_kPa": np.sqrt(mean_squared_error(y, predictions)),
        "R2": r2_score(y, predictions)
    })

    prediction_table[name] = predictions

scores_df = pd.DataFrame(scores).sort_values("RMSE_kPa")

print("CONTROLLED GPR EXPERIMENT")
display(scores_df.round(3))

print("\nPREDICTIONS BY FORMULATION")
display(prediction_table.round(3))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern

df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
).sort_values(
    ["GelMA_concentration_pct", "curing_time_s"]
).reset_index(drop=True)

X = df[[
    "GelMA_concentration_pct",
    "curing_time_s"
]].to_numpy()

y = df["mean_modulus_kPa"].to_numpy()

# Same starting experiments for every strategy
initial = [0, 5, 11]
total = len(y)

def fit_and_select(observed, beta=1.0):
    available = [
        i for i in range(total)
        if i not in observed
    ]

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X[observed])
    X_candidates = scaler.transform(X[available])

    log_y = np.log(y[observed])
    offset = log_y.mean()

    kernel = (
        ConstantKernel(
            1.0, constant_value_bounds="fixed"
        )
        * Matern(
            length_scale=[1.0, 1.0],
            length_scale_bounds="fixed",
            nu=2.5
        )
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        alpha=0.005,
        optimizer=None,
        normalize_y=False
    )

    gp.fit(X_train, log_y - offset)

    mu, sigma = gp.predict(
        X_candidates,
        return_std=True
    )

    acquisition = mu + offset + beta * sigma

    chosen_position = np.argmax(acquisition)

    return available[chosen_position]

# Sequential model-guided selection
observed = initial.copy()
ml_best = [max(y[observed])]
selection_log = []

while len(observed) < total:

    chosen = fit_and_select(observed)

    # Reveal the existing measurement only AFTER selection
    observed.append(chosen)

    selection_log.append({
        "experiment_number": len(observed),
        "GelMA_pct": X[chosen, 0],
        "curing_time_s": X[chosen, 1],
        "observed_modulus_kPa": y[chosen],
        "best_so_far_kPa": max(y[observed])
    })

    ml_best.append(max(y[observed]))

# Random selection baseline
rng = np.random.default_rng(42)

remaining = [
    i for i in range(total)
    if i not in initial
]

random_curves = []

for trial in range(1000):

    order = rng.permutation(remaining)
    chosen = initial.copy()
    best_values = [max(y[chosen])]

    for idx in order:
        chosen.append(int(idx))
        best_values.append(max(y[chosen]))

    random_curves.append(best_values)

random_curves = np.array(random_curves)

random_mean = random_curves.mean(axis=0)
random_low = np.percentile(random_curves, 5, axis=0)
random_high = np.percentile(random_curves, 95, axis=0)

steps = np.arange(len(initial), total + 1)

plt.figure(figsize=(9, 5))

plt.plot(
    steps, ml_best,
    marker="o",
    label="GPR-UCB selection"
)

plt.plot(
    steps, random_mean,
    linestyle="--",
    label="Mean random selection"
)

plt.fill_between(
    steps,
    random_low,
    random_high,
    alpha=0.15,
    label="Random 5th–95th percentile"
)

plt.axhline(
    y.max(),
    linestyle=":",
    label="Best measured formulation"
)

plt.xlabel("Number of formulations evaluated")
plt.ylabel("Best observed modulus (kPa)")
plt.title("Retrospective Active Learning: GelMA Design")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print("MODEL-GUIDED EXPERIMENT SELECTION")
display(pd.DataFrame(selection_log).round(3))

print("\nBest measured condition:")
display(df.loc[
    df["mean_modulus_kPa"].idxmax(),
    ["GelMA_concentration_pct",
     "curing_time_s",
     "mean_modulus_kPa"]
])
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from itertools import combinations

from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern

df = pd.read_csv(
    "data/processed/gelma_formulation_summary.csv"
).sort_values(
    ["GelMA_concentration_pct", "curing_time_s"]
).reset_index(drop=True)

X = df[[
    "GelMA_concentration_pct",
    "curing_time_s"
]].to_numpy()

y = df["mean_modulus_kPa"].to_numpy()

n = len(y)
best_idx = int(np.argmax(y))

def select_gpr_ucb(observed, beta=1.0):

    remaining = [
        i for i in range(n)
        if i not in observed
    ]

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X[observed])
    X_remaining = scaler.transform(X[remaining])

    log_y = np.log(y[observed])
    offset = log_y.mean()

    kernel = (
        ConstantKernel(
            1.0, constant_value_bounds="fixed"
        )
        * Matern(
            length_scale=[1.0, 1.0],
            length_scale_bounds="fixed",
            nu=2.5
        )
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        alpha=0.005,
        optimizer=None,
        normalize_y=False
    )

    gp.fit(X_train, log_y - offset)

    mu, std = gp.predict(
        X_remaining,
        return_std=True
    )

    acquisition = mu + offset + beta * std

    return remaining[int(np.argmax(acquisition))]

def experiments_to_optimum(initial, strategy, rng=None):

    observed = list(initial)

    if best_idx in observed:
        return len(observed)

    while best_idx not in observed:

        if strategy == "gpr":
            chosen = select_gpr_ucb(observed)

        elif strategy == "random":
            remaining = [
                i for i in range(n)
                if i not in observed
            ]
            chosen = int(rng.choice(remaining))

        observed.append(chosen)

    return len(observed)

# Every possible set of 3 initial formulations
starts = list(combinations(range(n), 3))

rng = np.random.default_rng(42)

benchmark = []

for initial in starts:

    gpr_count = experiments_to_optimum(
        initial, "gpr"
    )

    # Multiple random runs for each starting set
    random_counts = [
        experiments_to_optimum(
            initial, "random", rng
        )
        for _ in range(100)
    ]

    benchmark.append({
        "initial": str(initial),
        "gpr_experiments": gpr_count,
        "random_mean_experiments": np.mean(random_counts)
    })

benchmark_df = pd.DataFrame(benchmark)

print("TOTAL STARTING COMBINATIONS:", len(benchmark_df))

print("\nGPR average experiments:",
      benchmark_df["gpr_experiments"].mean())

print("Random average experiments:",
      benchmark_df["random_mean_experiments"].mean())

display(benchmark_df.head(15).round(2))

# Aggregate comparison
plt.figure(figsize=(7, 5))

plt.bar(
    ["GPR-UCB", "Random"],
    [
        benchmark_df["gpr_experiments"].mean(),
        benchmark_df["random_mean_experiments"].mean()
    ]
)

plt.ylabel("Average experiments to reach best formulation")
plt.title("Active-Learning Robustness Benchmark")
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ast import literal_eval

# Parse each initial set
evaluation = benchmark_df.copy()
evaluation["initial_indices"] = evaluation["initial"].apply(
    literal_eval
)

# Exclude starts that already contain the optimum
search_cases = evaluation[
    ~evaluation["initial_indices"].apply(
        lambda indices: best_idx in indices
    )
].copy()

# For 12 conditions and 3 initial observations:
# 9 remain untested, and the optimum is equally likely
# to appear at any position in a random ordering.
random_expected = 3 + (9 + 1) / 2

gpr_average = search_cases["gpr_experiments"].mean()

improvement = (
    (random_expected - gpr_average)
    / random_expected
) * 100

print("ACTIVE LEARNING — SEARCH-ONLY BENCHMARK")
print("=" * 55)

print("Starting combinations requiring search:", len(search_cases))
print(f"GPR mean total experiments: {gpr_average:.3f}")
print(f"Random expected total experiments: {random_expected:.3f}")
print(f"Relative reduction: {improvement:.2f}%")

# How often does the policy find the optimum
# within different total experiment budgets?
budget_results = []

for budget in range(3, 13):

    gpr_success = (
        search_cases["gpr_experiments"] <= budget
    ).mean()

    # Under uniform random search, the optimum is
    # equally likely to occupy any of 9 remaining positions.
    random_success = max(
        0, min((budget - 3) / 9, 1)
    )

    budget_results.append({
        "Total experiments": budget,
        "GPR success probability": gpr_success,
        "Random success probability": random_success
    })

budget_df = pd.DataFrame(budget_results)

display(budget_df.round(3))

plt.figure(figsize=(9, 5))

plt.plot(
    budget_df["Total experiments"],
    budget_df["GPR success probability"],
    marker="o",
    label="GPR-UCB"
)

plt.plot(
    budget_df["Total experiments"],
    budget_df["Random success probability"],
    marker="s",
    linestyle="--",
    label="Random selection"
)

plt.xlabel("Total formulations evaluated")
plt.ylabel("Fraction of searches finding the best formulation")
plt.title("Experimental Search Efficiency — 165 Starting Sets")
plt.ylim(0, 1.05)
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

from itertools import combinations
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern
from sklearn.preprocessing import StandardScaler

def choose_next(observed, strategy):
    remaining = [
        i for i in range(len(y))
        if i not in observed
    ]

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X[observed])
    X_candidates = scaler.transform(X[remaining])

    log_y = np.log(y[observed])
    offset = log_y.mean()

    kernel = (
        ConstantKernel(1.0, constant_value_bounds="fixed")
        * Matern(
            length_scale=[1.0, 1.0],
            length_scale_bounds="fixed",
            nu=2.5
        )
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        alpha=0.005,
        optimizer=None,
        normalize_y=False
    )

    gp.fit(X_train, log_y - offset)

    mu, sigma = gp.predict(
        X_candidates,
        return_std=True
    )

    if strategy == "Greedy":
        scores = mu + offset
    elif strategy == "Uncertainty":
        scores = sigma
    elif strategy == "GPR-UCB":
        scores = mu + offset + sigma
    else:
        raise ValueError("Unknown strategy")

    return remaining[int(np.argmax(scores))]


def run_search(initial, strategy):
    observed = list(initial)

    while best_idx not in observed:
        chosen = choose_next(observed, strategy)
        observed.append(chosen)

    return len(observed)


strategies = ["Greedy", "Uncertainty", "GPR-UCB"]

# Exclude cases where optimum is already known
starts = [
    initial
    for initial in combinations(range(len(y)), 3)
    if best_idx not in initial
]

records = []

for initial in starts:
    for strategy in strategies:
        count = run_search(initial, strategy)

        records.append({
            "initial": str(initial),
            "strategy": strategy,
            "experiments": count
        })

ablation = pd.DataFrame(records)

summary = (
    ablation.groupby("strategy")
    .agg(
        mean_experiments=("experiments", "mean"),
        median_experiments=("experiments", "median"),
        success_by_5=(
            "experiments",
            lambda s: (s <= 5).mean()
        ),
        success_by_6=(
            "experiments",
            lambda s: (s <= 6).mean()
        )
    )
    .reset_index()
)

# Exact random-search expectations
random_summary = pd.DataFrame([{
    "strategy": "Random (expected)",
    "mean_experiments": 8.0,
    "median_experiments": 8.0,
    "success_by_5": 2 / 9,
    "success_by_6": 3 / 9
}])

summary = pd.concat(
    [summary, random_summary],
    ignore_index=True
)

print("ACQUISITION STRATEGY COMPARISON")
display(summary.round(3))

plt.figure(figsize=(9, 5))

plt.bar(
    summary["strategy"],
    summary["mean_experiments"]
)

plt.ylabel("Mean total experiments to find optimum")
plt.title("Active Learning: Acquisition Function Ablation")
plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import re
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LinearRegression

folder = Path.cwd() / "GelMA_CompressionFiles"

records = []

for path in sorted(folder.glob("*.xlsx")):

    match = re.search(
        r"(\d+)p(\d+)s",
        path.name,
        flags=re.IGNORECASE
    )

    if not match:
        continue

    concentration = int(match.group(1))
    curing_time = int(match.group(2))

    workbook = pd.ExcelFile(path)

    for sheet in workbook.sheet_names:

        # Only numbered specimen sheets
        if not re.search(r"_\d+$", sheet):
            continue

        raw = pd.read_excel(
            path,
            sheet_name=sheet,
            header=None
        )

        header_rows = raw.index[
            raw.iloc[:, 0]
            .astype(str)
            .str.strip()
            .eq("Load (N)")
        ]

        if len(header_rows) == 0:
            continue

        start = header_rows[0] + 2

        data = raw.iloc[start:, [4, 5]].copy()
        data.columns = ["strain", "stress_Pa"]

        for col in data.columns:
            data[col] = pd.to_numeric(
                data[col],
                errors="coerce"
            )

        data = data.dropna()

        result = {
            "GelMA_pct": concentration,
            "curing_time_s": curing_time,
            "specimen": sheet
        }

        for upper in [0.03, 0.05]:

            region = data[
                data["strain"].between(
                    0.01, upper
                )
            ]

            column = f"modulus_1_{int(upper*100)}_kPa"

            if len(region) < 3:
                result[column] = np.nan
                continue

            model = LinearRegression()

            model.fit(
                region[["strain"]],
                region["stress_Pa"]
            )

            result[column] = model.coef_[0] / 1000

        records.append(result)

sensitivity = pd.DataFrame(records)

comparison = (
    sensitivity.groupby(
        ["GelMA_pct", "curing_time_s"]
    )[["modulus_1_3_kPa", "modulus_1_5_kPa"]]
    .mean()
    .reset_index()
)

# Read the reported means from the original worksheet
# Rows 4-15 correspond to the 12 formulation conditions.
reported_rows = raw_summary.iloc[4:16]

reported_means = pd.to_numeric(
    reported_rows.iloc[:, 15],
    errors="coerce"
).to_numpy()

if len(comparison) != len(reported_means):
    raise ValueError("Formulation counts do not match")

comparison["reported_mean_kPa"] = reported_means

comparison["error_1_3_kPa"] = (
    comparison["modulus_1_3_kPa"]
    - comparison["reported_mean_kPa"]
)

comparison["error_1_5_kPa"] = (
    comparison["modulus_1_5_kPa"]
    - comparison["reported_mean_kPa"]
)

print("MODULUS EXTRACTION COMPARISON")
display(comparison.round(3))

print("\nMean absolute discrepancy, 1-3%:")
print(comparison["error_1_3_kPa"].abs().mean())

print("\nMean absolute discrepancy, 1-5%:")
print(comparison["error_1_5_kPa"].abs().mean())
```

```python

from pathlib import Path
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

base = Path.cwd()
folder = base / "GelMA_CompressionFiles"

records = []

for path in sorted(folder.glob("*.xlsx")):

    match = re.search(
        r"(\d+)p(\d+)s",
        path.name,
        flags=re.IGNORECASE
    )

    if not match:
        continue

    concentration = int(match.group(1))
    curing_time = int(match.group(2))

    # Use the documented methodology
    if concentration == 20 and curing_time in [90, 120]:
        upper_strain = 0.03
    else:
        upper_strain = 0.06

    for sheet in pd.ExcelFile(path).sheet_names:

        specimen_match = re.search(r"_(\d+)$", sheet)

        if not specimen_match:
            continue

        specimen_number = int(specimen_match.group(1))

        raw = pd.read_excel(
            path,
            sheet_name=sheet,
            header=None
        )

        header_rows = raw.index[
            raw.iloc[:, 0]
               .astype(str)
               .str.strip()
               .eq("Load (N)")
        ]

        if len(header_rows) == 0:
            continue

        start = header_rows[0] + 2

        data = raw.iloc[start:, [4, 5]].copy()
        data.columns = ["strain", "stress_Pa"]

        for column in data.columns:
            data[column] = pd.to_numeric(
                data[column],
                errors="coerce"
            )

        data = data.dropna()

        region = data[
            data["strain"].between(0.01, upper_strain)
        ]

        if len(region) < 3:
            print("Insufficient points:", path.name, sheet)
            continue

        model = LinearRegression()
        model.fit(
            region[["strain"]],
            region["stress_Pa"]
        )

        records.append({
            "filename": path.name,
            "specimen": sheet,
            "specimen_number": specimen_number,
            "GelMA_concentration_pct": concentration,
            "curing_time_s": curing_time,
            "strain_upper_pct": upper_strain * 100,
            "modulus_kPa": model.coef_[0] / 1000,
            "fit_R2": model.score(
                region[["strain"]],
                region["stress_Pa"]
            ),
            "fit_points": len(region)
        })

corrected = pd.DataFrame(records)

corrected = corrected.sort_values([
    "GelMA_concentration_pct",
    "curing_time_s",
    "specimen_number"
]).reset_index(drop=True)

print("Corrected specimen records:", len(corrected))

print("Unique formulations:", corrected[
    ["GelMA_concentration_pct", "curing_time_s"]
].drop_duplicates().shape[0])

display(corrected.head(15).round(3))

output = base / "data" / "processed"
output.mkdir(parents=True, exist_ok=True)

corrected.to_csv(
    output / "gelma_modulus_corrected.csv",
    index=False
)
```

```python

source = base / "GelMA_Characterization_Data.xlsx"

original = pd.read_excel(
    source,
    sheet_name="YoungsModulus",
    header=None
)

reported_records = []

concentrations = [10, 15, 20]
times = [30, 60, 90, 120]

for row_idx, (concentration, curing_time) in enumerate(
    [(c, t) for c in concentrations for t in times],
    start=4
):
    for specimen_number in range(1, 13):

        value = pd.to_numeric(
            original.iloc[row_idx, specimen_number + 1],
            errors="coerce"
        )

        reported_records.append({
            "GelMA_concentration_pct": concentration,
            "curing_time_s": curing_time,
            "specimen_number": specimen_number,
            "reported_kPa": value
        })

reported_df = pd.DataFrame(reported_records)

validated = corrected.merge(
    reported_df,
    on=[
        "GelMA_concentration_pct",
        "curing_time_s",
        "specimen_number"
    ],
    how="outer",
    validate="one_to_one",
    indicator=True
)

if not (validated["_merge"] == "both").all():
    print("Unmatched records:")
    display(validated[validated["_merge"] != "both"])
    raise ValueError("Specimen matching is incomplete.")

validated["difference_kPa"] = (
    validated["modulus_kPa"] -
    validated["reported_kPa"]
)

validated["absolute_error_kPa"] = (
    validated["difference_kPa"].abs()
)

summary = validated.groupby([
    "GelMA_concentration_pct",
    "curing_time_s"
]).agg(
    specimens=("modulus_kPa", "count"),
    calculated_mean=("modulus_kPa", "mean"),
    reported_mean=("reported_kPa", "mean"),
    specimen_MAE=("absolute_error_kPa", "mean"),
    max_specimen_error=("absolute_error_kPa", "max")
).reset_index()

summary["mean_difference_kPa"] = (
    summary["calculated_mean"] -
    summary["reported_mean"]
)

print("CORRECTED MODULUS VALIDATION")
display(summary.round(3))

print("\nOverall specimen-level MAE (kPa):")
print(round(validated["absolute_error_kPa"].mean(), 3))

print("\nLargest individual discrepancies:")
display(
    validated.sort_values(
        "absolute_error_kPa",
        ascending=False
    )[[
        "specimen",
        "reported_kPa",
        "modulus_kPa",
        "difference_kPa",
        "fit_R2"
    ]].head(12).round(3)
)

validated.to_csv(
    output / "gelma_specimen_validation.csv",
    index=False
)
```

```python

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

path = (
    Path.cwd()
    / "GelMA_CompressionFiles"
    / "20p120sCompression.xlsx"
)

sheet = "20p120s_2"

raw = pd.read_excel(
    path,
    sheet_name=sheet,
    header=None
)

print("SPECIMEN METADATA")
display(raw.iloc[:7, :15])

# Identify the measurement header
headers = raw.index[
    raw.iloc[:, 0]
    .astype(str)
    .str.strip()
    .eq("Load (N)")
]

if len(headers) != 1:
    raise ValueError("Check measurement header location")

data = raw.iloc[headers[0] + 2:, :6].copy()

data.columns = [
    "load_N",
    "time_s",
    "extension_mm",
    "corrected_extension_mm",
    "strain",
    "stress_Pa"
]

for col in data.columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

data = data.dropna(
    subset=["strain", "stress_Pa"]
)

# Use the original documented 1-3% interval
region = data[
    data["strain"].between(0.01, 0.03)
].copy().sort_values("strain")

if len(region) < 3:
    raise ValueError("Not enough observations")

X = region[["strain"]]
y = region["stress_Pa"]

# Ordinary regression
ordinary = LinearRegression().fit(X, y)

# Regression through origin
origin = LinearRegression(
    fit_intercept=False
).fit(X, y)

# Secant using the closest available endpoints
low = region.iloc[0]
high = region.iloc[-1]

secant = (
    (high["stress_Pa"] - low["stress_Pa"]) /
    (high["strain"] - low["strain"])
) / 1000

reported = 160.411

methods = pd.DataFrame({
    "Method": [
        "Ordinary regression",
        "Through-origin regression",
        "Secant slope"
    ],
    "Modulus_kPa": [
        ordinary.coef_[0] / 1000,
        origin.coef_[0] / 1000,
        secant
    ]
})

methods["Difference_kPa"] = (
    methods["Modulus_kPa"] - reported
)

print("\nFITTING INTERVAL")
print("Points:", len(region))
print("Minimum strain:", region["strain"].min())
print("Maximum strain:", region["strain"].max())
print("Ordinary intercept (Pa):", ordinary.intercept_)

print("\nMETHOD COMPARISON")
display(methods.round(3))

# Plot original measurements and both fitted lines
strain_grid = np.linspace(
    region["strain"].min(),
    region["strain"].max(),
    100
)

grid = pd.DataFrame({"strain": strain_grid})

plt.figure(figsize=(9, 5))

plt.scatter(
    region["strain"] * 100,
    region["stress_Pa"] / 1000,
    label="Measured data",
    s=20
)

plt.plot(
    strain_grid * 100,
    ordinary.predict(grid) / 1000,
    label="Ordinary regression"
)

plt.plot(
    strain_grid * 100,
    origin.predict(grid) / 1000,
    label="Through-origin regression"
)

plt.xlabel("Compressive strain (%)")
plt.ylabel("Compressive stress (kPa)")
plt.title("GelMA 20%, 120 s — Specimen 2")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd

from sklearn.model_selection import (
    LeaveOneOut, cross_val_predict
)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import (
    StandardScaler, PolynomialFeatures
)
from sklearn.linear_model import Ridge
from sklearn.compose import TransformedTargetRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    "data/processed/gelma_source_formulations.csv"
)

X = df[[
    "GelMA_concentration_pct",
    "curing_time_s"
]]

y = df["mean_modulus_kPa"]

def log_ridge_model():
    return TransformedTargetRegressor(
        regressor=make_pipeline(
            StandardScaler(),
            PolynomialFeatures(
                degree=2,
                include_bias=False
            ),
            Ridge(alpha=1.0)
        ),
        func=np.log,
        inverse_func=np.exp
    )

models = {
    "Mean baseline": DummyRegressor(
        strategy="mean"
    ),
    "Log-target Quadratic Ridge": log_ridge_model()
}

results = []
predictions = df.copy()

for name, model in models.items():

    pred = cross_val_predict(
        model,
        X,
        y,
        cv=LeaveOneOut()
    )

    predictions[name] = pred

    results.append({
        "Model": name,
        "MAE_kPa": mean_absolute_error(y, pred),
        "RMSE_kPa": np.sqrt(
            mean_squared_error(y, pred)
        ),
        "LOOCV_R2": r2_score(y, pred)
    })

print("SOURCE-REPORTED MODULUS MODEL VALIDATION")

display(pd.DataFrame(results).round(3))

print("\nHELD-OUT FORMULATION PREDICTIONS")
display(predictions.round(3))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from itertools import combinations

from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern

# ==========================================
# 1. LOAD SOURCE-REPORTED EXPERIMENTAL DATA
# ==========================================

df = pd.read_csv(
    "data/processed/gelma_source_formulations.csv"
).sort_values(
    ["GelMA_concentration_pct", "curing_time_s"]
).reset_index(drop=True)

X = df[
    ["GelMA_concentration_pct", "curing_time_s"]
].to_numpy()

y = df["mean_modulus_kPa"].to_numpy()

assert len(df) == 12
assert np.all(y > 0)

best_idx = int(np.argmax(y))

print("BEST OBSERVED FORMULATION")
display(df.iloc[[best_idx]])

# ==========================================
# 2. DEFINE EXPERIMENT SELECTION
# ==========================================

def choose_next(observed, strategy):

    remaining = [
        i for i in range(len(y))
        if i not in observed
    ]

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X[observed])
    X_candidates = scaler.transform(X[remaining])

    log_y = np.log(y[observed])
    offset = log_y.mean()

    kernel = (
        ConstantKernel(
            1.0,
            constant_value_bounds="fixed"
        )
        * Matern(
            length_scale=[1.0, 1.0],
            length_scale_bounds="fixed",
            nu=2.5
        )
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        alpha=0.005,
        optimizer=None,
        normalize_y=False
    )

    gp.fit(X_train, log_y - offset)

    mu, sigma = gp.predict(
        X_candidates,
        return_std=True
    )

    if strategy == "Greedy":
        scores = mu + offset

    elif strategy == "Uncertainty":
        scores = sigma

    elif strategy == "GPR-UCB":
        scores = mu + offset + sigma

    else:
        raise ValueError(
            f"Unknown strategy: {strategy}"
        )

    return remaining[int(np.argmax(scores))]

# ==========================================
# 3. DEFINE THE SEARCH SIMULATION
# ==========================================

def run_search(initial, strategy):

    observed = list(initial)

    while best_idx not in observed:
        selected = choose_next(
            observed,
            strategy
        )

        observed.append(selected)

    return len(observed)

# ==========================================
# 4. RUN ALL STARTING COMBINATIONS
# ==========================================

strategies = [
    "Greedy",
    "Uncertainty",
    "GPR-UCB"
]

starts = [
    initial
    for initial in combinations(range(len(y)), 3)
    if best_idx not in initial
]

records = []

for initial in starts:

    for strategy in strategies:

        count = run_search(
            initial,
            strategy
        )

        records.append({
            "initial": str(initial),
            "strategy": strategy,
            "experiments": count
        })

source_ablation = pd.DataFrame(records)

# ==========================================
# 5. SUMMARIZE RESULTS
# ==========================================

source_summary = (
    source_ablation.groupby("strategy")
    .agg(
        mean_experiments=(
            "experiments", "mean"
        ),
        median_experiments=(
            "experiments", "median"
        ),
        success_by_5=(
            "experiments",
            lambda s: (s <= 5).mean()
        ),
        success_by_6=(
            "experiments",
            lambda s: (s <= 6).mean()
        )
    )
    .reset_index()
)

random_baseline = pd.DataFrame([{
    "strategy": "Random (expected)",
    "mean_experiments": 8.0,
    "median_experiments": 8.0,
    "success_by_5": 2/9,
    "success_by_6": 3/9
}])

source_summary = pd.concat(
    [source_summary, random_baseline],
    ignore_index=True
)

print("\nSOURCE-REPORTED ACTIVE LEARNING BENCHMARK")
print("Starting combinations:", len(starts))

display(source_summary.round(3))

# ==========================================
# 6. VISUALIZE THE BENCHMARK
# ==========================================

plt.figure(figsize=(9, 5))

plt.bar(
    source_summary["strategy"],
    source_summary["mean_experiments"]
)

plt.ylabel(
    "Mean total experiments to find optimum"
)

plt.title(
    "Active Learning Benchmark — Source-Reported GelMA Data"
)

plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()

# ==========================================
# 7. SAVE THE RESULTS
# ==========================================

output = Path("reports")
output.mkdir(parents=True, exist_ok=True)

source_ablation.to_csv(
    output / "source_active_learning_runs.csv",
    index=False
)

source_summary.to_csv(
    output / "source_active_learning_summary.csv",
    index=False
)

print("\nResearch benchmark saved successfully.")
```
